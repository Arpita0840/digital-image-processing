```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a random image
img = np.random.randint(0, 256, (300, 300), dtype=np.uint8)

# Apply Median Filter
filtered = cv2.medianBlur(img, 5)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Before Median Filter")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(filtered, cmap='gray')
plt.title("After Median Filter")
plt.axis('off')

plt.tight_layout()
plt.show()
```
