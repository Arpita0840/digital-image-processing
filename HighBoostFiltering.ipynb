```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a random image
img = np.random.randint(0, 256, (300, 300), dtype=np.uint8)

# Apply Gaussian Blur
blur = cv2.GaussianBlur(img, (5, 5), 0)

# High Boost Filtering
A = 1.5
filtered = cv2.addWeighted(img, A, blur, 1-A, 0)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Before High Boost Filtering")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(filtered, cmap='gray')
plt.title("After High Boost Filtering")
plt.axis('off')

plt.tight_layout()
plt.show()
```
