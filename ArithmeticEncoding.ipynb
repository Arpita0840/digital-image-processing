```python
# Arithmetic Encoding

def arithmetic_encoding(message):
    # Calculate probability of each character
    symbols = sorted(set(message))
    freq = {s: message.count(s) for s in symbols}

    total = len(message)
    low_range = 0.0
    high_range = 1.0

    # Calculate cumulative probabilities
    cum_prob = {}
    low = 0.0

    for s in symbols:
        prob = freq[s] / total
        cum_prob[s] = (low, low + prob)
        low += prob

    # Encode message
    for s in message:
        range_value = high_range - low_range
        high_range = low_range + range_value * cum_prob[s][1]
        low_range = low_range + range_value * cum_prob[s][0]

    code = (low_range + high_range) / 2

    print("Original Message:", message)
    print("Encoded Value:", code)
    print("Final Range:", (low_range, high_range))


# Input message
message = "ABAB"

arithmetic_encoding(message)
```
