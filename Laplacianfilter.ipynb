```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a random image
img = np.random.randint(0, 256, (300, 300), dtype=np.uint8)

# Apply Laplacian Filter
filtered = cv2.Laplacian(img, cv2.CV_64F)

# Convert result to displayable image
filtered = np.uint8(np.absolute(filtered))

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Before Laplacian Filter")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(filtered, cmap='gray')
plt.title("After Laplacian Filter")
plt.axis('off')

plt.tight_layout()
plt.show()
```
